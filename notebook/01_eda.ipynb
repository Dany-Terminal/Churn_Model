# How to deal with null columns
Since if we remove all the null col we will be reducing the 50k rows to just 17k, so thats not a good step.

We should try to fill as many cols as we can

```python
cols_with_null = df.columns[df.isnull().any()]
for col in cols_with_null:
    print(f'-----{col}----')
    print(f"Min: {df[col].min()}")
    print(f"Max: {df[col].max()}")
    print(f'median:{df[col].median()}')
    print(f'mean:{df[col].mean().round(0)}')
    print(f'25th percentil:{df[col].quantile(0.25)}')
    print(f'75th percentil:{df[col].quantile(0.75)}')
    print(f'IQR:{df[col].quantile(0.75) - df[col].quantile(0.25)}')
    print(f'Outlier: {((df[col] < df[col].quantile(0.25) - 1.5*(df[col].quantile(0.75)-df[col].quantile(0.25))) | (df[col] > df[col].quantile(0.75) + 1.5*(df[col].quantile(0.75)-df[col].quantile(0.25)))).sum()}')
```

# Taking a quick look in the detail of each col
**Age** have many dirty values and needs to be clean [18-85]


**Returns_Rate**,Right-swek-> customer return a lot


**Days_Since_Last_Purchase**, Right sweked->many inactive users

Below can be ignored as well (low skew)

 **Customer_Service_Calls**, slight skew-> may indicate unhappy customer

 
 **Product_Reviews_Written**, slight skew-> few user write review
 
Others may have a bit of swekness, but can be ignored

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
#coerce skip null to avoid error
df.loc[(df['Age']<18) | (df['Age']>85), 'Age'] = np.nan #convert impossible val to NAN
```

```python
df['Total_Purchases'] = pd.to_numeric(df['Total_Purchases'], errors='coerce')
#coerce skip null to avoid error
df.loc[(df['Total_Purchases']<0) | (df['Total_Purchases']>130), 'Total_Purchases'] = np.nan #convert impossible val to NAN
```

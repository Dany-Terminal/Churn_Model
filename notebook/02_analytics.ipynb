5. High Total Purchase -> less churn
6. Less customer service call -> Less Churn | **Strong indicator**
7. High Email Open rate -> Less Churn | **Strong indicator**
8. More product review -> Less Churn
9. Less Life time value -> Less Churn | **Imp to Look into**

## Possible Insigts on Life_time Value(LTV)

1. LTV may be calculated on based on profit, or the company have a free/freemium model that is used the most
2. After 5 customer service call, user mostlikely gets churn, (recommendation: A servey need to be done on what are these problem user are facing that arent being solved thus getting churned)
3. Many churns occure in LTV=0, indicating: User iseither using free tire or their returns aren't profitable comparing to what company is spending on them also could be just used to browse through product without buying.
4. LTV of ~[1000-2000] have least churned user
5. Churns are highly of areas where, 0 total purchase,irrespective of mebership years, indicating users are intrested in other benifits of membership (maybe a model like amazon, where membership gives amazon prime). But also indicates user getting churned, a possibility that the take mebership only when any offer/discount is avaliable, otherwise not.
6. Users having Total purchase >=20 , have very less chances of getting churned.
7. pattern incustomer service call and total purchase: Total purchases <=40 have higher chances of churning, ,for <=40 total purchase users customer service call >5 have slightly high chances of churning and >10have high chances of churning

**NOTE**

Its noticed that high LTV customers are being churned, they either need to be brought out of the free/freemium model or expenditure on free model needs to be cut.
